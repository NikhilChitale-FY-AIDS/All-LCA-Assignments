```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Loading the csv dataset
df = pd.read_csv("assignment 7.csv")

#Display the dataset
print("Dataset: ")
print(df)
```

```python
#Line Plot
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
# Bar Plot
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Math"])
plt.title("Math Marks of Students")
plt.xlabel("Student")
plt.ylabel("Math Marks")
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Math"], df["Python"], s=70)
plt.title("Math vs Python Marks")
plt.xlabel("Math Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
# Correlation Matrix and Heatmap
correlation = df[["Math", "Python", "Attendance"]].corr()
print("Correlation Matrix:")
print(correlation)

plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()

```

```python
# CHALLENGE / EXTENSION TASK

# 1.Histogram of Python marks
plt.figure(figsize=(7, 4))
plt.hist(df["Python"], bins=5, edgecolor="black", label="Python Marks")
plt.title("Histogram of Python Marks")
plt.xlabel("Python Marks")
plt.ylabel("Number of Students")
plt.legend()
plt.show()
```

```python
# 2.Box plot for Math and Python marks
plt.figure(figsize=(7, 4))
plt.boxplot(
    [df["Math"], df["Python"]],
    tick_labels=["Math", "Python"]
)
plt.title("Box Plot of Math and Python Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
# 3.Pie chart showing students from each city
city_count = df["City"].value_counts()

plt.figure(figsize=(7, 5))
plt.pie(
    city_count,
    labels=city_count.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Number of Students from Each City")
plt.legend(city_count.index, title="City")
plt.show()
```

```python
# 4.Calculate and display avg math marks,python marks and attendance
print("Average Math Marks:", df["Math"].mean())
print("Average Python Marks:", df["Python"].mean())
print("Average Attendance:", df["Attendance"].mean())
```
