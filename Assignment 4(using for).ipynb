```python
# Create two matrices using nested lists
matrix1 = [
    [1, 2, 3],
    [4, 5, 6]
]

matrix2 = [
    [7, 8, 9],
    [10, 11, 12]
]

# Initialize result matrix
result = [
    [0, 0, 0],
    [0, 0, 0]
]

# Add the two matrices
for i in range(len(matrix1)):
    for j in range(len(matrix1[0])):
        result[i][j] = matrix1[i][j] + matrix2[i][j]

# Display the result
print("Resultant Matrix:")
for row in result:
    print(row)
```
