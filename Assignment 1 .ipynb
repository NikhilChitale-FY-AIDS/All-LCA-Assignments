Exercise 1: Dictionary Operations

Algorithm / Pseudocode:
1. Start.
2. Create a dictionary named my_dict with 5 student details.
3. Display the student information.
4. Add a new student detail.
5. Display the updated dictionary.
6. Delete a student detail.
7. Display the dictionary after deletion.
8. Stop.


```python
my_dict = {
    "name": "ABC",
    "rollno": 34,
    "division": "2",
    "course": "Computer Engineering",
    "age": 20
}

print("Original Student Information:")
print(my_dict)

my_dict["college"] = "XYZ College"

print("\nAfter Adding Student Information:")
print(my_dict)

del my_dict["age"]

print("\nAfter Deleting Student Information:")
print(my_dict)
```

```python
# Exercise 2: Convert List to Dictionary

list1 = ['name', 'division', 'rollno']
list2 = ['ABC', '2', 34]

my_dict = dict(zip(list1, list2))

print("Dictionary:", my_dict)
```

```python
# Exercise 3: Sort Dictionary

my_dict = {
    'name': 'ABC',
    'divisoion': '2',
    'rollno': 34,
    'marks': [65, 87, 67, 94]
}

sorted_dict = sorted(my_dict)

print("Sorted elements of dictionary:")
print(sorted_dict)
```

```python
# Exercise 4: Convert Dictionary to Lists

my_dict = {
    'name': 'ABC',
    'division': '2',
    'rollno': 34
}

keys_list = list(my_dict.keys())
values_list = list(my_dict.values())

print("List of keys:", keys_list)
print("List of values:", values_list)
```

```python
# Exercise 5: Mean of Dictionary Values

mydict = {
    'marks1': 23,
    'marks2': 123,
    'marks3': 43,
    'marks4': 13,
    'marks5': 39
}

values = mydict.values()

mean = sum(values) / len(values)

print("Mean of all values:", mean)
```

```python
# Exercise 7: Frequency of Letters

string = input("Enter a string: ")

frequency = {}

for letter in string:
    if letter in frequency:
        frequency[letter] += 1
    else:
        frequency[letter] = 1

print("Frequency of letters:")
print(frequency)
```

```python
# Exercise 8: Quotient and Remainder

numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

result = (quotient, remainder)

print("Tuple of Quotient and Remainder:", result)
```
