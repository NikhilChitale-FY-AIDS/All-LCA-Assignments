```python
# Assignment - 2
# Greatest Number

num_1 = float(input("Enter First Number:"))
num_2 = float(input("Enter Second Number:"))
num_3 = float(input("Enter Third Number:"))

if num_1 >= num_2 and num_1 >= num_3:
    print("Greatest Number =", num_1)

elif num_2 >= num_1 and num_2 >= num_3:
    print("Greatest Number =" , num_2)

else:
    print("Greatest Number =", num_3)
    
```

```python
# Determining Grade

x = float(input("Enter Your Mathematics Marks: "))

if 100 >= x >= 90:
    print("Grade Assigned - O")

elif 89 >= x >= 80:
    print("Grade Assigned - A+")

elif 79 >= x >= 70:
    print("Grade Assigned - A")      

elif 69 >= x >= 60:
    print("Grade Assigned - B")

elif 59 >= x >= 50:
    print("Grade Assigned - C")

elif 49 >= x >= 40:
    print("Grade Assigned - P")

elif x == 39:
    print("You are between P and F")

elif x > 100:
    print("Stop Lying")

elif x < 40:
    print("Grade Assigned - F")          

```
